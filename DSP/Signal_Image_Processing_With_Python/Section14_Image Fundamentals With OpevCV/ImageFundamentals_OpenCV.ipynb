# Images Flipping Along Axes

```python
# Image Flipping along X-axis

img_flipX = cv2.flip(img_RGB,0)
plt.imshow(img_flipX)
plt.show()
```

```python
# Image Flipping along Y-axis

img_flipY = cv2.flip(img_RGB,1)
plt.imshow(img_flipY)
plt.show()
```

```python
# Image Flipping along both X and Y-axes

img_flipXY = cv2.flip(img_RGB, -1)
plt.imshow(img_flipXY)
plt.show()
```

```python
plt.figure(figsize=(12,8))
plt.subplot(2,2,1)
plt.imshow(img_RGB)
plt.title("Original Image", fontsize = 15)
plt.subplot(2,2,2)
plt.imshow(img_flipX)
plt.title("Flipped Along X-axis", fontsize = 15)
plt.subplot(2,2,3)
plt.imshow(img_flipY)
plt.title("Filpped Along Y-axis", fontsize = 15)
plt.subplot(2,2,4)
plt.imshow(img_flipXY)
plt.title("Flipped Along Both X and Y axes", fontsize = 15)
plt.show()
```
