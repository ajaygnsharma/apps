# Histogram of colour Images

```python
dark = cv2.imread('DarkGreen.png')
```

```python
plt.figure(figsize=(16,8))
plt.imshow(dark)
plt.show()
```

```python
dark_RGB = cv2.cvtColor(dark, cv2.COLOR_BGR2RGB) # to display image using matplotlib, we have converetd the color order from
                                                  # BGR To RGB
plt.figure(figsize=(16,8))
plt.imshow(dark_RGB)
plt.show()
```

```python
print(dark_RGB.shape)
print(" ")
print(np.max(dark_RGB))
print(" ")
print(np.min(dark_RGB))
```

### OpenCV Histogram

**cv2.calcHist(images, channels, mask, histSize, ranges[, hist[, accumulate]])**

* images : it is the source image that should be specified in square bracket[image].
* channels : it is zero for gray scale. For color image, it is [0], [1] or [2] to calculate histogram of blue, green or red channel respectively. It shouls also be specified in square bracket [0]. 
* mask : mask image. To find histogram of full image, it is given as “None”. 
* histSize : It represents the BIN count. Need to be specified in square brackets. For full scale, its value is [256].
* ranges : this is our RANGE. Normally, it is [0,256].

```python
hist_values = cv2.calcHist([dark_RGB],channels = [0], mask = None, histSize = [256], ranges = [0,256]) # RGB
```

```python
hist_values.shape
```

```python
plt.figure(figsize=(14,7))
plt.plot(hist_values, "r")
plt.title("Histogram of Red Color Only", fontsize = 15)
plt.xlabel("Intensity Levels", fontsize = 15)
plt.ylabel("Frequency", fontsize = 15)
plt.show()
```

```python
hist_values = cv2.calcHist([dark_RGB],channels = [1], mask = None, histSize = [256], ranges = [0,256]) # RGB
plt.figure(figsize=(14,7))
plt.plot(hist_values, "g")
plt.title("Histogram of Green Color Only", fontsize = 15)
plt.xlabel("Intensity Levels", fontsize = 15)
plt.ylabel("Frequency", fontsize = 15)
plt.show()
```

```python
hist_values = cv2.calcHist([dark_RGB],channels = [2], mask = None, histSize = [256], ranges = [0,256]) # RGB
plt.figure(figsize=(14,7))
plt.plot(hist_values, "b")
plt.title("Histogram of Blue Color Only", fontsize = 15)
plt.xlabel("Intensity Levels", fontsize = 15)
plt.ylabel("Frequency", fontsize = 15)
plt.show()
```

```python
# Viewing Separate Color Channels
color = ('r', 'g', 'b')
plt.figure(figsize=(14,7)) 
# We now separate the colors and plot each in the Histogram
for i, col in enumerate(color):
    color_hist = cv2.calcHist([dark_RGB], [i], None, [256], [0, 256])
    plt.plot(color_hist, color = col)
    plt.xlim([-1,256])
    plt.xlabel("Intensity Levels", fontsize = 15)
    plt.ylabel("Frequency", fontsize = 15)
   

plt.show()
```

# Histogram Equalization

```python
skull = cv2.imread('Skull.jpg', 0)
```

```python
plt.figure(figsize=(12,6))
plt.imshow(skull, cmap = 'gray')
plt.show()
```

```python
skull_equalized = cv2.equalizeHist(skull) # Applying histogram equalization
```

```python
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.imshow(skull, cmap = 'gray')
plt.xlabel("Intensity Levels", fontsize = 15)
plt.ylabel("Frequency", fontsize = 15)
plt.title("Input Image", fontsize = 15)
plt.subplot(1,2,2)
plt.imshow(skull_equalized, cmap = 'gray')
plt.xlabel("Intensity Levels", fontsize = 15)
plt.ylabel("Frequency", fontsize = 15)
plt.title("Histogram Equalized Image", fontsize = 15)
plt.show()
```

```python
# Histogram of inout Image
hist_values_skull = cv2.calcHist([skull],channels = [0], mask = None, histSize = [256], ranges = [0,256])
```

```python
# Histogram of Equalized Image
hist_values = cv2.calcHist([skull_equalized],channels=[0],mask=None,histSize=[256],ranges=[0,256]) # channels[0] coz gray scale
```

```python
plt.figure(figsize=(20,7))
plt.subplot(1,2,1)
plt.plot(hist_values_skull,'m')
plt.xlabel("Intensity Levels", fontsize = 15)
plt.ylabel("Frequency", fontsize = 15)
plt.title("Histogram of Input Image", fontsize = 15)
plt.subplot(1,2,2)
plt.plot(hist_values)
plt.ylim([0,12200])
plt.xlabel("Intensity Levels", fontsize = 15)
plt.ylabel("Frequency", fontsize = 15)
plt.title("Histogram of Processed Image", fontsize = 15)
plt.show()
```

# Histogram Equalization for Color Image

```python
barbara = cv2.imread('Barbara.jpg')
```

```python
barbara_RGB = cv2.cvtColor(barbara , cv2.COLOR_BGR2RGB) 
plt.figure(figsize=(12,6))
plt.imshow(barbara_RGB)
plt.show()
```

```python
barbara_HSV = cv2.cvtColor(barbara, cv2.COLOR_BGR2HSV)
```

```python
# Observing V channel for applying Histogram Equalization

barbara_HSV[:,:,2]
```

```python
barbara_HSV[:,:,2] = cv2.equalizeHist(barbara_HSV[:,:,2])  # Applying Histogram Equalization on V channel
```

```python
barbara_equalized = cv2.cvtColor(barbara_HSV, cv2.COLOR_HSV2RGB)
```

```python
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.imshow(barbara_RGB)
plt.title("Input Image", fontsize = 15)
plt.subplot(1,2,2)
plt.imshow(barbara_equalized)
plt.title("Histogram Equalized Image", fontsize = 15)
plt.show()
```
