```python
import sympy as sym

t = sym.symbols('t', real=True)

# scalar constant
b = 1

# the signals:
x = sym.cos(t)
y = sym.cos(t)+b

p1 = sym.plot(x, (t, 0, 2*sym.pi), legend=True, label='cos(t)', show=False);
p2 = sym.plot(y, (t, 0, 2*sym.pi), legend=True, label='cos(t)+b', show=False);

p1.append(p2[0])
p1.show()
```

```python
# scalar constant
a = 0.5

# the signal:
x = sym.cos(t)
y = a*sym.cos(t)

p1 = sym.plot(x, (t, 0, 2*sym.pi), legend=True, label='cos(t)', show=False);
p2 = sym.plot(y, (t, 0, 2*sym.pi), legend=True, label='a cos(t)', show=False);

p1.append(p2[0])
p1.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(11)  # Discrete time indices
x = np.array([1, 2, 3, 4, 5, 4, 3, 2, 1, 0, 0])  # Signal values

plt.stem(n, x)
plt.xlabel('n')
plt.ylabel('x[n]')
plt.title('Discrete Signal Representation in Python')
plt.grid(True)
plt.show()
```

```python
from scipy.signal import decimate
import numpy as np

x_t = np.array([0,1,2,3,4]);
x_2t = decimate(x_t, 2, ftype='fir');
print(x_2t)
```

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.array([0,1,2,3,4]);
x_t = np.array([0,1,2,3,4]);

x_t_c = x_t[::2];
x_t_full = np.zeros_like(x_t);
x_t_full[::2] = x_t_c;
plt.stem(t, x_t_full);
plt.show();
```

```python
import scipy.interpolate as interp

t_orig = np.array([0, 1, 2, 3, 4])
x_t = np.array([0, 1, 2, 3, 4])
a = 2

# Interpolate compressed values
t_new = np.linspace(0, 4, len(x_t) // a + 1)  # New time points
interp_func = interp.interp1d(t_orig[::a], x_t[::a], kind="linear", fill_value="extrapolate")
x_t_interpolated = interp_func(t_orig)

# Plot
plt.figure(figsize=(6, 4))
plt.stem(t_orig, x_t, markerfmt='bo', linefmt="b-", label="Original Signal")
plt.stem(t_orig, x_t_interpolated, markerfmt='ro', linefmt="r-", label="Compressed Signal x(2t)")
plt.xlabel('Time (t)')
plt.ylabel('x(t)')
plt.title('Signal Compression with Interpolation')
plt.legend()
plt.grid(True)
plt.show()
```
