Integration of linear functions: Frame 12

```python
integrate((3*x+2)**4,x)
```

```python
u = symbols('u');
u_expr = 3*x+2
du_dx = 3
integrand_in_u = (u**4)/du_dx
integral_in_u = integrate(integrand_in_u, u)
integral_final = integral_in_u.subs(u, u_expr)
```

```python
print(integral_final)
```
