```python
n = symbols('n',integer=True);
x_expr = n**2 + 5;
k = symbols('k', integer=True);
x_shifted = x_expr.subs(n,n-k);
print(x_shifted)
```

```python
x_seq = [x_shifted.subs(n,i) for i in range(5)]
print(x_seq)
```

```python
for i in range(5):
    print(x_seq[i].subs(k,i));
    
```
